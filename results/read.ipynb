```python
import pandas as pd
```

```python
df = pd.read_csv("layer_sweep_results.csv")
```

```python
df = df.sort_values(by=['layer'])
```

```python
df['layer'] = df['layer'] + 33
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].bar(df['layer'], df['precision'])
axes[0].set_xlabel('Layer')
axes[0].set_ylabel('Precision')
axes[0].set_title('Precision by Layer')
axes[0].set_ylim(0.65, 0.85)
max_precision_idx = df['precision'].idxmax()
axes[0].scatter(df.loc[max_precision_idx, 'layer'], df.loc[max_precision_idx, 'precision'], 
                color='red', s=100, zorder=5, marker='*')
axes[0].set_title(f'Precision by Layer (Max: {df.loc[max_precision_idx, "precision"]:.4f} at Layer {df.loc[max_precision_idx, "layer"]})')

axes[1].bar(df['layer'], df['recall'])
axes[1].set_xlabel('Layer')
axes[1].set_ylabel('Recall')
axes[1].set_ylim(0.65, 0.85)
max_recall_idx = df['recall'].idxmax()
axes[1].scatter(df.loc[max_recall_idx, 'layer'], df.loc[max_recall_idx, 'recall'], 
                color='red', s=100, zorder=5, marker='*')
axes[1].set_title(f'Recall by Layer (Max: {df.loc[max_recall_idx, "recall"]:.4f} at Layer {df.loc[max_recall_idx, "layer"]})')

axes[2].bar(df['layer'], df['f1'])
axes[2].set_xlabel('Layer')
axes[2].set_ylabel('F1')
axes[2].set_ylim(0.65, 0.85)
max_f1_idx = df['f1'].idxmax()
axes[2].scatter(df.loc[max_f1_idx, 'layer'], df.loc[max_f1_idx, 'f1'], 
                color='red', s=100, zorder=5, marker='*')
axes[2].set_title(f'F1 by Layer (Max: {df.loc[max_f1_idx, "f1"]:.4f} at Layer {df.loc[max_f1_idx, "layer"]})')

plt.tight_layout()
plt.show()
```
