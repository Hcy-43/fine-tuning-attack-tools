```python
from pathlib import Path
import pandas as pd

base_dir = Path("baseline")

all_dfs = []

for folder in base_dir.iterdir():
    summary_path = folder / "GPT_result" / "summary.md"
    if summary_path.exists():
        df = pd.read_csv(summary_path, sep='|', skipinitialspace=True)
        df = df.dropna(axis=1, how='all').drop(0).reset_index(drop=True)
        df.columns = df.columns.str.strip()
        df = df.replace(r'\*\*', '', regex=True)  # 去掉粗體標記
        df["ratio"] = folder.name.split(sep='_')[-1]  # 加上來源資料夾欄位
        all_dfs.append(df)

# 合併成一個大 DataFrame
merged_df = pd.concat(all_dfs, ignore_index=True)
print(merged_df)
category_df = merged_df[merged_df['Category']=='Overall ']
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Filter for ratio 0.3 only
category_df_03 = category_df[category_df['ratio'] == '0.3'].copy()
category_df_lr_03 = category_df_lr[category_df_lr['ratio'] == '0.3'].copy()

# Convert columns to numeric
category_df_03['Score=5 Ratio'] = pd.to_numeric(category_df_03['Score=5 Ratio'])
category_df_lr_03['Score=5 Ratio'] = pd.to_numeric(category_df_lr_03['Score=5 Ratio'])

# Get categories
categories = category_df_03['Category'].values
num_vars = len(categories)

# Compute angle for each axis
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]  # complete the circle

# Create the plot
fig, ax = plt.subplots(figsize=(10, 10), subplot_kw=dict(projection='polar'))

# Plot baseline (category_df)
values_baseline = category_df_03['Score=5 Ratio'].astype(float).tolist()
values_baseline += values_baseline[:1]
ax.plot(angles, values_baseline, 'o-', linewidth=2, label='baseline (ratio=0.3)')
ax.fill(angles, values_baseline, alpha=0.1)

# Plot pca_lr (category_df_lr)
values_lr = category_df_lr_03['Score=5 Ratio'].astype(float).tolist()
values_lr += values_lr[:1]
ax.plot(angles, values_lr, 's-', linewidth=2, label='pca_lr (ratio=0.3)')
ax.fill(angles, values_lr, alpha=0.1)

# Fix axis to go in the right order and start at 12 o'clock
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)

# Set labels
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories)
ax.set_ylim(0, 1)
ax.legend(loc='upper right', bbox_to_anchor=(1.1, 1.1))
ax.set_title('Radar Plot Comparison (ratio=0.3)', size=16, y=1.08)
ax.grid(True)

plt.tight_layout()
plt.show()
```
