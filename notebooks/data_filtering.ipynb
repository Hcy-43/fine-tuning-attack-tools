```python
import pandas as pd
```

```python
train_benign = pd.read_csv("../data/beavertails/round0/30k/train/benign.csv")
train_harmful = pd.read_csv("../data/beavertails/round0/30k/train/harmful.csv")
```

```python
print("benign len: ", len(train_benign))
print("harmful len: ", len(train_harmful))
```

```python
def create_balanced_dataset(num, harmful_ratio, seed=42):
    """
    Create a balanced dataset with specified number of entries and harmful ratio.
    
    Args:
        num: Total number of entries in the output dataframe
        harmful_ratio: Proportion of harmful examples (0 to 1)
        seed: Random seed for reproducibility
        
    Returns:
        DataFrame with mixed benign and harmful examples
    """
    num_harmful = int(num * harmful_ratio)
    num_benign = num - num_harmful
    
    # Sample from each dataset
    sampled_harmful = train_harmful.sample(n=num_harmful, random_state=seed)
    sampled_benign = train_benign.sample(n=num_benign, random_state=seed)
    
    # Combine and shuffle
    combined_df = pd.concat([sampled_harmful, sampled_benign], ignore_index=True)
    combined_df = combined_df.sample(frac=1, random_state=seed).reset_index(drop=True)
    
    return combined_df
```
